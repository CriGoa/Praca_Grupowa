## Opis zbioru danych: Dane_A

Zbiór danych pochodzi z bazy FAOSTAT, prowadzonej przez Organizację Narodów Zjednoczonych do spraw Wyżywienia i Rolnictwa (FAO). Zawiera informacje o cenach producentów dla różnych produktów rolnych w Polsce, obejmując zarówno ceny nominalne (w lokalnej walucie PLN, standardowej lokalnej walucie SLC oraz w USD), jak i indeksy cen producenta (z bazą 2014–2016 = 100). Dane dotyczą produktów takich jak jabłka, morele, szparagi oraz kategorie zagregowane (np. strączkowe, warzywa, bulwy i korzenie).

Zbiór obejmuje rekordy roczne dla każdego produktu i rodzaju ceny w okresie od 1991 do 2024 roku. Łącznie zawiera kilkaset rekordów (dokładna liczba zależy od pełnego zbioru, ale przesłany fragment sugeruje dane dla wielu produktów i elementów). Każdy rekord obejmuje metadane, takie jak kody kraju (ISO3, M49), kody produktów (FAO, CPC), jednostki (LCU, SLC, USD lub brak dla indeksu) oraz wartości cen wraz z flagami jakości danych (np. A dla oficjalnych, I dla imputowanych).

Zakres czasowy różni się w zależności od produktu – np. dane dla jabłek zaczynają się od 1991 roku, a dla szparagów od 1996 roku. Dane dla 2024 roku są oznaczone flagą I, co wskazuje na prognozy lub wartości szacunkowe.
